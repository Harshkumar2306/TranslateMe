## 1. Sinusoidal Positional Encoding Visualization
Transformers do not use recurrence or convolution. Positional encodings provide order awareness:
$$\text{PE}_{(pos, 2i)} = \sin(pos / 10000^{2i/d_{model}})$$
$$\text{PE}_{(pos, 2i+1)} = \cos(pos / 10000^{2i/d_{model}})$$

```python
pe = PositionalEncoding(d_model=128, max_len=100, dropout=0.0)
pe_matrix = pe.pe.squeeze(0).numpy()

plt.figure(figsize=(10, 6))
plt.imshow(pe_matrix, aspect='auto', cmap='viridis')
plt.title('Sinusoidal Positional Encoding (d_model=128, max_len=100)')
plt.xlabel('Embedding Dimension')
plt.ylabel('Token Position')
plt.colorbar(label='Encoding Value')
plt.show()
```

## 2. Scaled Dot-Product Attention
$$\text{Attention}(Q, K, V) = \text{softmax}\left(\frac{QK^T}{\sqrt{d_k}} + M\right)V$$

```python
q = torch.randn(1, 4, 8, 16)
k = torch.randn(1, 4, 8, 16)
v = torch.randn(1, 4, 8, 16)

attention = ScaledDotProductAttention(dropout=0.0)
out, weights = attention(q, k, v)
print('Attention Output Shape:', out.shape)
print('Attention Weights Shape:', weights.shape)
```

## 3. Sequence-to-Sequence Transformer Forward Pass

```python
model = Seq2SeqTransformer(
    src_vocab_size=100,
    tgt_vocab_size=100,
    d_model=128,
    n_heads=4,
    num_encoder_layers=2,
    num_decoder_layers=2,
    d_ff=256
).to(device)

src = torch.randint(1, 100, (2, 10)).to(device)
tgt = torch.randint(1, 100, (2, 8)).to(device)
out = model(src, tgt)
print('Forward Pass Output Logits:', out.shape)
```
